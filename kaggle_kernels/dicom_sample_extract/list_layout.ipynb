```python
from pathlib import Path
BASE = Path("/kaggle/input/competitions/rsna-knee-abnormality-detection")
print(f"listing {BASE} ...")
for p in sorted(BASE.iterdir()):
    print(f"{p.name}  ({'dir' if p.is_dir() else 'file'})")
    if p.is_dir():
        children = sorted(p.iterdir())
        for c in children[:10]:
            print(f"  {c.name}  ({'dir' if c.is_dir() else 'file'})")
        if len(children) > 10:
            print(f"  ... ({len(children)} entries total)")
    elif p.suffix == ".csv":
        head = p.read_text().split('\n')[:3]
        print(f"    head: {head[0][:200]}")
        print(f"          {head[1][:200] if len(head)>1 else ''}")
```
